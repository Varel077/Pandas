```python
import pandas as pd
import os

csv_path = input("Masukkan path lengkap file titanicfull.csv: ").strip()

if not os.path.isfile(csv_path):
    raise FileNotFoundError("File tidak ditemukan. Path salah.")

df = pd.read_csv(csv_path)

df_backup = df.copy(deep=True)

df.columns = (
    df.columns
      .astype(str)
      .str.lower()
      .str.strip()
      .str.replace(' ', '_')
)

df.head()
```
